```python
from aocd import get_data
raw_data = get_data(day=1, year=2016)
data = [[instruction[0], int(instruction[1:])] for instruction in raw_data.split(", ")]
```

## Part 1

```python
pos = "north"
pos_dict = {}

mapping = {"R": {"north": "east", "east":"south", "south":"west", "west":"north"}, 
           "L": {"north": "west", "east":"north", "south":"east", "west":"south"}
}

for move, steps in data:
    pos = mapping[move][pos]
    pos_dict[pos] = pos_dict.get(pos, 0) + steps

abs(pos_dict.get("north", 0) - pos_dict.get("south", 0)) + abs(pos_dict.get("east", 0) - pos_dict.get("west", 0))
```

## Part 2

```python
def find_twice_visited_location(data: list) -> int:

    pos = "north"
    coord = [0, 0]
    locations = {(0, 0)}

    direction_steps = {
        "north": (0, 1),
        "south": (0, -1),
        "east": (1, 0),
        "west": (-1, 0),
    }

    for move, steps in data:
        pos = mapping[move][pos]
        dx, dy = direction_steps[pos]

        for _ in range(steps):
                coord[0] += dx
                coord[1] += dy

                current_location = tuple(coord)
                if current_location in locations:
                    return abs(coord[0]) + abs(coord[1])
                else: 
                    locations.add(current_location)

find_twice_visited_location(data)
```
