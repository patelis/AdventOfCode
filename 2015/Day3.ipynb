```python
from aocd import get_data
raw_data = get_data(day=3, year=2015)
```

```python
raw_data
```

## Part 1

```python
def santa_travel(locations: set, path: str | list):

    position = [0, 0]

    moves = {"^": {"dx":0, "dy":1}, "v": {"dx":0, "dy":-1},
             "<": {"dx":-1, "dy":0}, ">": {"dx":1, "dy":0}
             }

    for move in path:
        next_move = moves[move]
        position[0] += next_move["dx"]
        position[1] += next_move["dy"]

        locations.add(tuple(position))

    return None

locations = {(0,0)}
santa_travel(locations, path=raw_data)

len(locations)
```

## Part 2

```python
santa_path = [raw_data[i] for i in range(len(raw_data)) if i % 2]
robo_santa_path = [raw_data[i] for i in range(len(raw_data)) if not i % 2]

locations = {(0,0)}
santa_travel(locations, path=santa_path)
santa_travel(locations, path=robo_santa_path)

len(locations)
```
