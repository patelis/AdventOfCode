```python
from aocd import get_data
raw_data = get_data(day=2, year=2015)
data = [[int(coord) for coord in dims.split("x")] for dims in raw_data.splitlines()]
```

## Part 1

```python
surfaces = [[math.prod(dims) / dim for dim in dims] for dims in data]
gift_wraps = [ 2 * sum(dims) +  min(dims) for dims in  surfaces]
sum(gift_wraps)
```

## Part 2

```python
bows = sum([math.prod(dims) for dims in data])
ribbon = sum([2 * sum(sorted(dims)[:-1])  for dims in data])
bows + ribbon
```
