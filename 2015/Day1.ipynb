```python
from aocd import get_data
raw_data = get_data(day=1, year=2015)
```

## Part 1

```python
from collections import Counter
counter = Counter(raw_data)
counter["("] - counter[")"]
```

## Part 2

```python
floor = 0

for step, move in enumerate(raw_data):
    if move == "(":
        floor += 1
    else:
        floor -= 1

    if floor == -1:
        break

step + 1
```
