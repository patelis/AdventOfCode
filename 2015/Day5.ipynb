```python
from aocd import get_data
from collections import Counter
raw_data = get_data(day=5, year=2015)
data = raw_data.splitlines()
```

## Part 1

```python
matches = 0

for s in data:

    rule_counter = 1
    counter = Counter(s+"aeiou")
    vowel_counts = counter["a"] + counter["e"] + counter["i"] + counter["o"] + counter["u"] - 5
    if vowel_counts >= 3:
        rule_counter += 1

    for i in range(len(s)-1):
        if s[i] == s[i+1]:
            rule_counter += 1
            break

    if ("ab" in s) or ("cd" in s) or ("pq" in s) or ("xy" in s):
        rule_counter -= 1

    if rule_counter == 3:
        matches += 1

matches    
```

## Part 2

```python
matches = 0

for s in data:
    rule_counter = 0
    for i in range(len(s)-2):
        pattern = s[i] + s[i+1]
        if pattern in s[i+2:]:
            rule_counter += 1
            break
    
    for i in range(len(s)-2):
        if s[i] == s[i+2]:
            rule_counter += 1
            break
    
    if rule_counter == 2:
        matches += 1

matches    
```
